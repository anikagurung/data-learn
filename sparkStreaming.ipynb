```python
df = spark.readStream.format("delta").table("emp") 
```

```python
%sql
insert into emp values(7,'abc'),(8,'xyz'); 
```

```python
import uuid

def foreach_batch_function(batch_df, epoch_id):
    output_path = (
        "/Volumes/anikadatabricks/default/mydatalake/stream/output1/"
        + str(uuid.uuid4())
    )

    batch_df.write.format("json").mode("append").save(output_path)

query = (
    df.writeStream
    .foreachBatch(foreach_batch_function)
    .option(
        "checkpointLocation",
        "/Volumes/anikadatabricks/default/mydatalake/stream/checkpoint1/"
    )
    .trigger(availableNow=True)
    .start()
)

query.awaitTermination()
```
